### Understanding word embeddings
Word embeddings are numerical representations of words that capture their meaings and relationships. Words with similar meanings are represented by vectors that are close to each other in a matematical space. They are widely used in natural language processing for semantic search, text classification, recomendation systems, and RAG applications.

**Transform Architecture**

Encoder and Decoder

In machine learning, especially in **Transformer architectures**, the encoder and decoder have different roles.

Encoder

The **encoder** processes the input data and transforms it into a meaningful numerical representation.

For text, the process is usually:

```text
Text → Tokens → Embeddings → Encoder → Contextual Representation
```

The encoder analyzes the relationships between words or tokens and tries to understand their context and meaning.

It is commonly used for tasks such as:

- Text classification
- Sentiment analysis
- Semantic search
- Embedding generation

Decoder

The **decoder** is mainly responsible for generating output.

It produces the result **token by token**, using previous tokens and the information available from the model.

For example:

```text
Input: "What is the capital of Brazil?"

Decoder output:
"Brasília" → "is" → "the" → "capital" → "of" → "Brazil"
```

Decoder-based models are commonly used for:

- Text generation
- Chatbots
- Code generation
- Question answering

#### Encoder vs. Decoder


| Component | Main Function |
|---|---|
| **Encoder** | Understands and represents the input |
| **Decoder** | Generates the output |


A simple way to remember it is:

> **Encoder = understands the input**  
> **Decoder = generates the output**

**Transformers vs. LLMs**

A **Transformer** is a neural network architecture designed to process sequences of data, especially text. It uses mechanisms such as **self-attention** to understand relationships between words or tokens.

An **LLM (Large Language Model)** is a large model trained on massive amounts of text to understand and generate language.

Most modern LLMs are built using the **Transformer architecture**.

> **Transformer = architecture**  
> **LLM = large language model built with that architecture**

Examples of LLMs include **GPT, Llama, Gemini, and Claude**.
